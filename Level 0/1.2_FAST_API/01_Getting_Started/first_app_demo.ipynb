# FastAPI First App Demo

This notebook shows the smallest useful FastAPI app.

````python
from fastapi import FastAPI
````

````python
app = FastAPI(title="First FastAPI Demo")

@app.get("/")
def home():
    return {"message": "FastAPI is running"}

@app.get("/health")
def health():
    return {"status": "ok"}
````

Save the same code into `main.py` and run:

```bash
uvicorn main:app --reload
```
