# Sync vs Async Demo

This notebook compares a blocking route with a non-blocking route in FastAPI.

````python
import asyncio
import time
from fastapi import FastAPI

app = FastAPI(title="Sync vs Async Demo")
````

````python
@app.get("/sync-wait")
def sync_wait():
    start = time.perf_counter()
    time.sleep(2)
    duration = round(time.perf_counter() - start, 2)
    return {"mode": "sync", "waited_seconds": duration}

@app.get("/async-wait")
async def async_wait():
    start = time.perf_counter()
    await asyncio.sleep(2)
    duration = round(time.perf_counter() - start, 2)
    return {"mode": "async", "waited_seconds": duration}
````

Save the same code into `main.py` and run:

```bash
uvicorn main:app --reload
```

Then compare:

* `GET /sync-wait`
* `GET /async-wait`

The output shape is similar, but the async route uses a non-blocking wait.
