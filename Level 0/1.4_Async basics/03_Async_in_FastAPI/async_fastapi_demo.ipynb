# Async in FastAPI Demo

This notebook shows simple async endpoints in FastAPI.

```python
import asyncio
from fastapi import FastAPI
from pydantic import BaseModel

app = FastAPI(title="Async in FastAPI Demo")
```

```python
class ChatRequest(BaseModel):
    question: str
    document_id: int
```

```python
@app.get("/async-health")
async def async_health():
    await asyncio.sleep(0.2)
    return {"status": "ok", "mode": "async"}

@app.post("/chat")
async def chat(payload: ChatRequest):
    await asyncio.sleep(1)
    return {
        "status": "success",
        "document_id": payload.document_id,
        "answer": f"Async response for: {payload.question}"
    }
```

This is a demo pattern. In a real backend, the `await` points may represent async database, retrieval, or provider calls.
