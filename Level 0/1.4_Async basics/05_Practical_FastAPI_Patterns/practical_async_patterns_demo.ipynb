# Practical Async Patterns Demo

This notebook shows a small GenAI-style FastAPI route with async helper functions.

```python
import asyncio
from fastapi import FastAPI
from pydantic import BaseModel

app = FastAPI(title="Practical Async Patterns Demo")
```

```python
class SummaryRequest(BaseModel):
    question: str
    document_id: int
```

```python
async def fetch_document(document_id: int):
    await asyncio.sleep(0.8)
    return {"document_id": document_id, "content": "Quarterly growth was positive."}

async def fetch_prompt_settings():
    await asyncio.sleep(0.4)
    return {"style": "concise"}

async def call_llm(question: str, content: str, style: str):
    await asyncio.sleep(1)
    return f"[{style}] Answer for '{question}' based on: {content}"
```

```python
@app.post("/summaries")
async def create_summary(payload: SummaryRequest):
    document, settings = await asyncio.gather(
        fetch_document(payload.document_id),
        fetch_prompt_settings()
    )

    answer = await call_llm(
        question=payload.question,
        content=document["content"],
        style=settings["style"]
    )

    return {
        "status": "success",
        "document_id": payload.document_id,
        "answer": answer
    }
```

This is a useful beginner pattern for RAG, summarization, and document Q and A services built with FastAPI.
