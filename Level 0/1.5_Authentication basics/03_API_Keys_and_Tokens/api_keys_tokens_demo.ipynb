# API Keys and Tokens Demo

This notebook shows simple patterns for checking an API key and a bearer token.

```python
from fastapi import FastAPI, Header, HTTPException

app = FastAPI(title="API Keys and Tokens Demo")

EXPECTED_API_KEY = "demo-key-123"
EXPECTED_BEARER_TOKEN = "demo-token-abc"
```

```python
@app.get("/api-key-protected")
def api_key_protected(x_api_key: str = Header(default="")):
    if x_api_key != EXPECTED_API_KEY:
        raise HTTPException(status_code=401, detail="Invalid API key")
    return {"status": "success", "auth_type": "api_key"}
```

```python
@app.get("/token-protected")
def token_protected(authorization: str = Header(default="")):
    expected_header = f"Bearer {EXPECTED_BEARER_TOKEN}"
    if authorization != expected_header:
        raise HTTPException(status_code=401, detail="Invalid bearer token")
    return {"status": "success", "auth_type": "bearer_token"}
```

In real systems, credentials should come from secure config, not hardcoded demo values.
