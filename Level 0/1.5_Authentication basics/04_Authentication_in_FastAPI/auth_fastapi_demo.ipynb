# Authentication in FastAPI Demo

This notebook shows a simple dependency-based auth pattern.

```python
from fastapi import Depends, FastAPI, Header, HTTPException

app = FastAPI(title="Authentication in FastAPI Demo")
EXPECTED_API_KEY = "demo-key-123"
```

```python
def verify_api_key(x_api_key: str = Header(default="")):
    if x_api_key != EXPECTED_API_KEY:
        raise HTTPException(status_code=401, detail="Invalid API key")
    return {"client_name": "notes-demo-client"}
```

```python
@app.get("/health")
def health():
    return {"status": "ok"}

@app.get("/protected-summary")
def protected_summary(client = Depends(verify_api_key)):
    return {
        "status": "success",
        "client": client,
        "message": "You can now access the protected GenAI endpoint"
    }
```

This pattern is small, but it shows the core idea: keep auth logic reusable and separate from route logic.
