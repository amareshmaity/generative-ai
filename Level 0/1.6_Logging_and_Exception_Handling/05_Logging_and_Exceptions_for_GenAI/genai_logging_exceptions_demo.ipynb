# GenAI Logging and Exceptions Demo

This notebook shows a small multi-step GenAI-style flow with logs and controlled errors.

```python
import logging

logging.basicConfig(level=logging.INFO, format="%(levelname)s:%(message)s")
logger = logging.getLogger(__name__)
```

```python
def retrieve_document(document_id: int):
    logger.info(f"Retrieving document_id={document_id}")
    if document_id != 1:
        raise FileNotFoundError("Document not found")
    return "Demo document content"
```

```python
def answer_question(question: str, document_id: int):
    try:
        logger.info("Received question-answer request")
        document = retrieve_document(document_id)
        logger.info("Calling demo model provider")
        return {
            "status": "success",
            "answer": f"Demo answer for: {question}",
            "document_preview": document[:20]
        }
    except FileNotFoundError as exc:
        logger.warning(f"RAG flow failed: {exc}")
        return {"status": "error", "message": "Requested document is missing"}
```
