# GenAI Config Demo

This notebook shows a simple settings pattern for a GenAI backend.

```python
import os
```

```python
class Settings:
    openai_api_key = os.getenv("OPENAI_API_KEY")
    default_model = os.getenv("DEFAULT_MODEL", "gpt-4o-mini")
    embedding_model = os.getenv("EMBEDDING_MODEL", "text-embedding-3-small")
    top_k = os.getenv("TOP_K", "5")
```

```python
settings = Settings()

print(settings.default_model)
print(settings.embedding_model)
print(settings.top_k)
print("OPENAI_API_KEY available" if settings.openai_api_key else "OPENAI_API_KEY missing")
```
