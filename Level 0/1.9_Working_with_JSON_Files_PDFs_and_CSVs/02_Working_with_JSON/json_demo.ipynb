# JSON Demo

This notebook shows simple JSON serialization and parsing in Python.

```python
import json
```

```python
payload = {
    "question": "What is RAG?",
    "document_id": 5,
    "temperature": 0.2
}

json_text = json.dumps(payload, indent=2)
print(json_text)
```

```python
parsed = json.loads(json_text)
print(parsed)
print(parsed["question"])
```
