# Files and CSV Demo

This notebook shows simple file writing, file reading, and CSV handling.

```python
sample_text = "Hello from the backend notes"
```

```python
with open("demo_notes.txt", "w", encoding="utf-8") as file:
    file.write(sample_text)
```

```python
with open("demo_notes.txt", "r", encoding="utf-8") as file:
    print(file.read())
```

```python
import csv

rows = [
    {"name": "Alice", "score": 95},
    {"name": "Bob", "score": 88}
]

with open("demo_scores.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.DictWriter(file, fieldnames=["name", "score"])
    writer.writeheader()
    writer.writerows(rows)
```

```python
with open("demo_scores.csv", "r", encoding="utf-8") as file:
    reader = csv.DictReader(file)
    for row in reader:
        print(row)
```
