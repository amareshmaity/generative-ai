# Dependencies Demo

```python
from fastapi import Depends, FastAPI

app = FastAPI()
```

```python
def common_params(limit: int = 10, page: int = 1):
    return {"limit": limit, "page": page}

@app.get("/items")
def get_items(params: dict = Depends(common_params)):
    return params
```

In real apps, dependencies are often used for auth, config, and database access.
