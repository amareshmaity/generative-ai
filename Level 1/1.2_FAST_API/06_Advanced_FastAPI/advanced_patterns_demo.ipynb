# Advanced FastAPI Patterns Demo

```python
from fastapi import BackgroundTasks, FastAPI
from time import perf_counter

app = FastAPI()
```

```python
def write_log(message: str):
    with open('fastapi_demo.log', 'a', encoding='utf-8') as f:
        f.write(message + '\n')

@app.post('/jobs')
def create_job(background_tasks: BackgroundTasks):
    background_tasks.add_task(write_log, 'Background task executed')
    return {'message': 'Job accepted'}
```

You can extend this idea with middleware, security, startup events, and async provider calls.
