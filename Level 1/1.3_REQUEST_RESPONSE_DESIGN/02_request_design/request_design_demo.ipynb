# Request Design Demo

This notebook shows how request structure maps into a FastAPI app.

````python
from fastapi import FastAPI, Query
from pydantic import BaseModel, Field

app = FastAPI(title="Request Design Demo")
````

````python
class SummaryRequest(BaseModel):
    question: str = Field(..., min_length=5, description="User question or instruction")
    document_id: int = Field(..., gt=0)
    temperature: float = Field(default=0.2, ge=0.0, le=1.0)
````

````python
@app.get("/documents/{document_id}")
def get_document(document_id: int, include_chunks: bool = Query(default=False)):
    return {
        "document_id": document_id,
        "include_chunks": include_chunks
    }

@app.post("/summaries")
def create_summary(payload: SummaryRequest):
    return {
        "received_question": payload.question,
        "document_id": payload.document_id,
        "temperature": payload.temperature
    }
````

Save the same code into `main.py` and run:
