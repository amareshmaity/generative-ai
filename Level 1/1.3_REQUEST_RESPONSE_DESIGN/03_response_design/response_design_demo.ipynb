# Response Design Demo

This notebook shows how to return clear and predictable FastAPI responses.

```python
from fastapi import FastAPI, status
from pydantic import BaseModel
from typing import List

app = FastAPI(title="Response Design Demo")
```

```python
class DocumentItem(BaseModel):
    id: int
    title: str

class DocumentListResponse(BaseModel):
    items: List[DocumentItem]
    page: int
    total: int

class SummaryResponse(BaseModel):
    status: str
    document_id: int
    summary: str
```

```python
@app.get("/documents", response_model=DocumentListResponse)
def list_documents(page: int = 1):
    items = [
        {"id": 1, "title": "Quarterly Report"},
        {"id": 2, "title": "Customer Feedback"}
    ]
    return {
        "items": items,
        "page": page,
        "total": len(items)
    }

@app.post("/summaries", response_model=SummaryResponse, status_code=status.HTTP_201_CREATED)
def create_summary():
    return {
        "status": "success",
        "document_id": 1,
        "summary": "The report shows stable revenue growth."
    }
```

Notice how each endpoint returns a stable response shape. This makes the API easier for frontend and testing code to consume.
