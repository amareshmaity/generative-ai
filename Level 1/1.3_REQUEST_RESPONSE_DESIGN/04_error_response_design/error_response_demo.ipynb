# Error Response Demo

This notebook shows how FastAPI can return useful error responses for missing resources and invalid input.

```python
from fastapi import FastAPI, HTTPException
from pydantic import BaseModel, Field

app = FastAPI(title="Error Response Demo")
```

```python
class ChatRequest(BaseModel):
    question: str = Field(..., min_length=3)
    document_id: int = Field(..., gt=0)
```

```python
DOCUMENTS = {
    1: "Quarterly revenue report",
    2: "Support ticket summary"
}

@app.get("/documents/{document_id}")
def get_document(document_id: int):
    if document_id not in DOCUMENTS:
        raise HTTPException(
            status_code=404,
            detail={
                "code": "DOCUMENT_NOT_FOUND",
                "message": f"Document {document_id} does not exist"
            }
        )
    return {"document_id": document_id, "title": DOCUMENTS[document_id]}

@app.post("/chat")
def chat(payload: ChatRequest):
    if payload.document_id not in DOCUMENTS:
        raise HTTPException(
            status_code=404,
            detail={
                "code": "DOCUMENT_NOT_FOUND",
                "message": f"Document {payload.document_id} does not exist"
            }
        )
    return {
        "status": "success",
        "answer": f"Answer generated from document {payload.document_id}."
    }
```

Try sending a request with a missing `document_id` or a short `question` to see FastAPI validation and custom error handling together.
