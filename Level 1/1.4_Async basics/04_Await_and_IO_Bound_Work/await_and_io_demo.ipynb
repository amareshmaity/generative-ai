# Await and I O Demo

This notebook shows how async helper functions can be combined inside a FastAPI endpoint.

```python
import asyncio
from fastapi import FastAPI

app = FastAPI(title="Await and I O Demo")
```

```python
async def fetch_document(document_id: int):
    await asyncio.sleep(1)
    return {"document_id": document_id, "title": "Annual Report"}

async def fetch_user_preferences(user_id: int):
    await asyncio.sleep(1)
    return {"user_id": user_id, "tone": "concise"}
```

```python
@app.get("/context/{document_id}")
async def get_context(document_id: int, user_id: int = 1):
    document, prefs = await asyncio.gather(
        fetch_document(document_id),
        fetch_user_preferences(user_id)
    )
    return {
        "document": document,
        "preferences": prefs
    }
```

This pattern is useful when multiple independent I O operations are needed before building the final response.
