# GenAI Authentication Demo

This notebook illustrates the difference between user auth and provider credentials.

```python
def verify_user_token(token: str):
    if token != "user-token-demo":
        raise ValueError("Invalid user token")
    return {"user_id": 101, "plan": "pro"}

def call_model_provider(prompt: str, provider_api_key: str):
    return {
        "provider_called": True,
        "prompt": prompt,
        "provider_key_used": provider_api_key[:4] + "..."
    }
```

```python
def answer_question(user_token: str, prompt: str):
    user = verify_user_token(user_token)
    provider_api_key = "sk-demo-provider-key"
    provider_response = call_model_provider(prompt, provider_api_key)
    return {
        "user": user,
        "answer": "Demo response generated through backend-controlled provider access",
        "provider_response": provider_response
    }
```

The key lesson is that the backend owns the provider credential, while the user only proves identity to the backend.
