# Logging Basics Demo

This notebook shows simple Python logging examples.

```python
import logging

logging.basicConfig(level=logging.INFO, format="%(levelname)s:%(message)s")
```

```python
logging.info("Application started")
logging.warning("Using demo configuration")
logging.error("Document retrieval failed for document_id=42")
```

```python
def summarize_document(document_id: int):
    logging.info(f"Starting summarization for document_id={document_id}")
    if document_id != 1:
        logging.error(f"Document not found for document_id={document_id}")
        return {"status": "error"}
    logging.info(f"Summarization completed for document_id={document_id}")
    return {"status": "success"}
```
