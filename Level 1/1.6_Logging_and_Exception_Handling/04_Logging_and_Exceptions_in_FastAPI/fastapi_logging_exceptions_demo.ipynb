# FastAPI Logging and Exceptions Demo

This notebook shows a simple API route with logging and `HTTPException`.

```python
import logging
from fastapi import FastAPI, HTTPException

logging.basicConfig(level=logging.INFO, format="%(levelname)s:%(message)s")
logger = logging.getLogger(__name__)
app = FastAPI(title="FastAPI Logging and Exceptions Demo")
```

```python
@app.get("/documents/{document_id}")
def get_document(document_id: int):
    logger.info(f"Fetching document_id={document_id}")
    if document_id != 1:
        logger.warning(f"Document not found for document_id={document_id}")
        raise HTTPException(status_code=404, detail="Document not found")
    logger.info(f"Document retrieval succeeded for document_id={document_id}")
    return {"document_id": document_id, "status": "success"}
```

This is a small pattern, but it reflects how many real API routes should behave.
