# GenAI Testing Demo

This notebook shows a small prompt-building function and a mocked provider-style call.

```python
def build_prompt(question: str, context: str) -> str:
    return f"Answer the question using the context.\nContext: {context}\nQuestion: {question}"
```

```python
def fake_model_call(prompt: str) -> dict:
    return {"answer": "Demo answer", "prompt_used": prompt}
```

```python
def test_build_prompt_contains_question_and_context():
    prompt = build_prompt("What is RAG?", "RAG combines retrieval with generation.")
    assert "What is RAG?" in prompt
    assert "RAG combines retrieval with generation." in prompt
```

```python
def test_fake_model_call_returns_answer_shape():
    result = fake_model_call("demo prompt")
    assert "answer" in result
    assert result["answer"] == "Demo answer"
```
